```python
AverageGoals_df = pd.read_csv('Data/Average_Goals.csv')

x = AverageGoals_df['Average temp']
y = AverageGoals_df['Average goals']

plt.scatter(x, y, color='blue', label='Data Points')

m, b = np.polyfit(x, y, 1)  
plt.plot(x, m*x + b, color='red', label='Line of Best Fit')

custom_y_ticks = [1,1.5,2,2.5,3,3.5,4,4.5,5,5.5,6]
plt.yticks(custom_y_ticks)

plt.xlabel('Average Temperature (°C)')
plt.ylabel('Average Goals per Game per Season')
plt.title('Average Goals vs Temperature')
plt.legend()

plt.show()
```

```python
AverageGoals_df = pd.read_csv('Data/Average_Goals.csv')

top_20 = AverageGoals_df.sort_values(by='Average goals', ascending=False).head(20)

top_20['Year'] = top_20['Season']

top_20 = top_20.set_index('Year')

plt.figure(figsize=(10, 8))
top_20['Average goals'].sort_values(ascending=True).plot(kind='barh', color='mediumseagreen')

plt.xlabel('Average Goals per Season')
plt.ylabel('Season')
plt.title('Top 20 Highest Scoring Seasons (Sorted by Goals)')

plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
