```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

AverageAttendance_df = pd.read_csv('/users/josh/desktop/720056188/Data/Average_Attendance.csv')
Rainfall_df = pd.read_csv('/users/josh/desktop/720056188/Data/Rainfall.csv')
Temperature_df = pd.read_csv('/users/josh/desktop/720056188/Data/Temperature.csv')

Attendance_Weather_df = pd.merge(AverageAttendance_df, Rainfall_df, on='date')
Attendance_Weather_df = pd.merge(Attendance_Weather_df, Temperature_df, on='date')

Attendance_Weather_df = Attendance_Weather_df.drop(columns=['1961 to 1990 long-term average_x', '1961 to 1990 long-term average_y', '1991 to 2020 long-term average_x', '1991 to 2020 long-term average_y', 'UK trend', 'Annual mean temperature (trend)'])
Attendance_Weather_df = Attendance_Weather_df.rename(columns={'UK': 'Annual Rainfall'})

Attendance_Weather_df.to_csv('/users/josh/desktop/720056188/Data/Attendance_Weather.csv', index=False)

Attendance_Weather_df.head(5)
```

```python
x = Attendance_Weather_df['date']
Attendance = Attendance_Weather_df['Average Attendance']
Rainfall = Attendance_Weather_df['Annual Rainfall']

fig, ax1 = plt.subplots(figsize=(10, 6))

ax1.set_xlabel('Season')
ax1.set_ylabel('Average Attendance per Game')
ax1.plot(x, Attendance, color= 'blue', marker = '+', linestyle='--', label='Attendance')
ax1.tick_params(axis='y', labelcolor='blue')
ax1.set_xticks(x)
ax1.set_xticklabels(x, rotation = 30)
plt.legend(loc = 'upper left')

ax2 = ax1.twinx()  
ax2.set_ylabel('Average Rainfall (mm)')  
ax2.plot(x, Rainfall, color='red', marker = 'D', label='Rainfall')
ax2.tick_params(axis='y', labelcolor='red')
plt.legend(loc = 'upper right')

fig.tight_layout()  
plt.title('Average Attendance vs Rainfall')

plt.show()
```

```python
x = Attendance_Weather_df['date']
Attendance = Attendance_Weather_df['Average Attendance']
Temperature = Attendance_Weather_df['Annual mean temperature']

fig, ax1 = plt.subplots(figsize=(10, 6))

ax1.set_xlabel('Season')
ax1.set_ylabel('Average Attendance per Game')
ax1.plot(x, Attendance, color= 'blue', marker = '+', linestyle='--', label='Attendance')
ax1.tick_params(axis='y', labelcolor='blue')
ax1.set_xticks(x)
ax1.set_xticklabels(x, rotation = 30)
plt.legend(loc = 'upper left')

ax2 = ax1.twinx()  
ax2.set_ylabel('Average Temperature (°C)')  
ax2.plot(x, Temperature, color='Green', marker = 'D', label='Temperature')
ax2.tick_params(axis='y', labelcolor='Green')
plt.legend(loc = 'upper right')

fig.tight_layout()  
plt.title('Average Attendance vs Temperature')

plt.show()
```
