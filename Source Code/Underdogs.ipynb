```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

Underdogs_df = pd.read_csv('/users/josh/desktop/720056188/Data/Underdogs_Weather.csv')

Underdogs_df.head()
```

```python
Sun_Total = 0
Rain_Total = 0
Cloud_Total = 0

for index, row in Underdogs_df.iterrows():
    if row['Weather'] == 'Sun':
        Sun_Total += 1
    elif row['Weather'] == 'Rain':
        Rain_Total += 1
    elif row['Weather'] == 'Cloud':
        Cloud_Total += 1

Weather_Total = [Sun_Total, Rain_Total, Cloud_Total]
labels=['Sun', 'Rain', 'Cloud']
colors = ['Yellow', '#87CEEB', 'Grey']

plt.pie(Weather_Total,labels=['Sun', 'Rain', 'Cloud'], colors = colors, autopct='%1.1f%%', startangle=140)
plt.title('Weather Distribution of Big Upsets')

plt.show()
```

```python
x = Underdogs_df['Average Temp']
y = Underdogs_df['Game']

plt.scatter(x, y, color='blue', label='Data Points')

m, b = np.polyfit(x, y, 1)  
plt.plot(x, m*x + b, color='red', label='Line of Best Fit')

custom_y_ticks = [1,1.5,2,2.5,3,3.5,4,4.5,5,5.5,6]
plt.yticks(custom_y_ticks)

plt.xlabel('Average Temperature (°C)')
plt.ylabel('Average Goals per Game per Season')
plt.title('Average Goals vs Temperature')
plt.legend()

plt.show()
```
